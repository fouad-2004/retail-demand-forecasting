```python
import pandas as pd

from forecasting import (
    clean_transactions,
    build_weekly_sales,
    walk_forward_forecasts,
    score_forecasts
)

raw = pd.read_excel("data/Online Retail.xlsx")
sales_clean = clean_transactions(raw)

print("Cleaned rows:", len(sales_clean))
```

```python
week_ends = pd.date_range(
    "2010-12-12",
    "2011-12-04",
    freq="W-SUN"
)

weekly = build_weekly_sales(
    sales_clean,
    product_code="85123A",
    week_ends=week_ends,
    unknown_weeks=["2011-01-02"]
)

cutoff = pd.Timestamp("2011-10-03")

train = weekly.loc[weekly.index < cutoff]
validation_dates = train.index[-8:]
test_dates = weekly.index[weekly.index >= cutoff]
```

```python
validation_predictions = walk_forward_forecasts(
    train, validation_dates
)

test_predictions = walk_forward_forecasts(
    weekly, test_dates
)

validation_scores = score_forecasts(validation_predictions)
test_scores = score_forecasts(test_predictions)

metrics = (
    validation_scores["MAE"]
    .rename("ValidationMAE")
    .to_frame()
    .join(test_scores["MAE"].rename("TestMAE"))
)

display(metrics.round(2))
```
